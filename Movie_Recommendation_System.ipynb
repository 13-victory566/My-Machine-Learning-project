```python
# Importing Dependencies
import pandas as pd
import difflib
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Loading the Dataset
Movie_data=pd.read_csv(r'C:\Users\victory\Desktop\MACHINE LEARNING PROJECTS\movies.csv', encoding='latin-1')
Movie_data['index']=Movie_data.index
Movie_data['title']=Movie_data['title'].str.lower()
print(Movie_data)
```

```python
# Combining features
Combined_features= Movie_data['title']+''+Movie_data['genres']
```

```python
# Vectorize Combined Feature
Vectorizer=TfidfVectorizer()
Vectorized_feature=Vectorizer.fit_transform(Combined_features)
```

```python
# Find the similarities between features
Similarity_score=cosine_similarity(Vectorized_feature)
```

```python
Movie=input('Enter Movie name:').lower()
```

```python
list_of_all_movies=Movie_data.title.str.lower()
list_of_all_movies.values
```

```python
# Finding close Match Movies
find_close_match=difflib.get_close_matches(Movie,list_of_all_movies)
print(find_close_match)
```

```python
close_match=find_close_match[0]
index_movie=Movie_data[Movie_data.title==close_match]['index'].values[0]
print(index_movie)
```

```python
indices=np.argsort(Similarity_score[index_movie])[::-1]
max_similarity_score=Similarity_score[index_movie][indices]
max_similarity_score
```

```python
m=indices.shape[0]
Recommended_movies=[]
for i in range(m):
  Related_movies=Movie_data[Movie_data['index']==indices[i]]['title'].values[0]
  Recommended_movies.append(Related_movies)
selected_movies=pd.Series(Recommended_movies)
print(selected_movies.head(30))
```
