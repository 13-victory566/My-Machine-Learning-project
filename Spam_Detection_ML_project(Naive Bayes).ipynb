```python
# Loading Dependencies
import pandas as pd
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Loading the Dataset
data=pd.read_csv(r'C:\Users\victory\Desktop\MACHINE LEARNING PROJECTS\spam (1).csv', encoding='Latin-1')
data['v2']=data['v2'].str.lower()

# Data Splitting
X=data['v2']
Y=data['v1']
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)

# Data pipeline Preprocessing

preprocessing=Pipeline([('tfidf', CountVectorizer())])
model=Pipeline([('preprocessing', preprocessing), ('classifier',MultinomialNB())])

# Training
model.fit(X_train,Y_train)
y_pred=model.predict(X_test)
y_train_pred=model.predict(X_train)

# Evaluation
print(f'ACCURACY SCORE FOR TEST DATA: {accuracy_score(Y_test,y_pred)}')
print(f'ACCURACY SCORE FOR TRAINING DATA: {accuracy_score(Y_train,y_train_pred)}')
print(f'CROSS VALIDATION SCORE: {cross_val_score(model,X_train,Y_train,cv=5)}')
print(f'CLASSIFICATION REPORT:\n{classification_report(Y_test,y_pred)}')
print(f'CONFUSION MATRIX:\n{confusion_matrix(Y_test,y_pred)}')
```
